```python
"""
Appends this week's BO inputs and outputs to the respective .npy files
in initial_data_2/function_N/.

Run from the directory that contains the initial_data_2 folder:
    python append_bo_data.py
"""

import numpy as np
from pathlib import Path

# ── New data ──────────────────────────────────────────────────────────────────
new_inputs = {
    1: [0.078579, 0.396923],
    2: [0.721449, 0.941278],
    3: [0.285526, 0.298043, 0.383102],
    4: [0.497411, 0.437024, 0.392893, 0.499691],
    5: [0.265828, 0.785871, 0.956868, 0.901489],
    6: [0.407502, 0.433135, 0.348846, 0.699075, 0.125416],
    7: [0.012807, 0.405906, 0.232567, 0.322556, 0.241670, 0.664147],
    8: [0.101730, 0.236262, 0.086905, 0.401978, 0.711456, 0.286539, 0.003732, 0.557067],
}

new_outputs = {
    1: 1.7560674160531887e-83,
    2: 0.6073429195218324,
    3: -0.02800851495609476,
    4: -2.337431233035072,
    5: 1410.822472306069,
    6: -0.47055730749435426,
    7: 1.8860590228330538,
    8: 9.7967762336111,
}

# ── Append ────────────────────────────────────────────────────────────────────
base = Path("initial_data_2")

for fn in range(1, 9):
    folder = base / f"function_{fn}"
    inp_path = folder / "initial_inputs.npy"
    out_path = folder / "initial_outputs.npy"

    # Load existing
    X = np.load(inp_path)
    Y = np.load(out_path)

    print(f"\nFunction {fn}:")
    print(f"  Inputs  before: {X.shape}  -> ", end="")

    # Append new row
    x_new = np.array(new_inputs[fn]).reshape(1, -1)
    y_new = np.array([new_outputs[fn]])

    X_updated = np.vstack([X, x_new])
    Y_updated = np.append(Y, y_new)

    print(f"{X_updated.shape}")
    print(f"  Outputs before: {Y.shape}  -> {Y_updated.shape}")
    print(f"  New x: {x_new.ravel()}")
    print(f"  New y: {y_new[0]:.6e}")

    # Save back
    np.save(inp_path, X_updated)
    np.save(out_path, Y_updated)

print("\nDone — all files updated.")
```
